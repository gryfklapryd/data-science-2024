```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
produk = ['Roti', 'Selai', 'Susu', 'Sereal', 'Telur', 'Keju', 'Kopi', 'Gula', 'Teh', 'Mentega']
# Buat 50 transaksi, tiap transaksi berisi 2-5 produk
transaksi = []
for _ in range(50):
  n_item = np.random.randint(2, 6)
  transaksi.append(list(np.random.choice(produk, n_item, replace=False)))

  # Suntikkan pola: Roti sering bersama Selai
for i in range(0, 20):
  if 'Roti' in transaksi[i] and 'Selai' not in transaksi[i]:
    transaksi[i].append('Selai')

  print('Contoh transaksi:', transaksi[:3])
  print('Jumlah transaksi:', len(transaksi))
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(transaksi).transform(transaksi)
df = pd.DataFrame(te_ary, columns=te.columns_)
print(df.head())
```

```python
from mlxtend.frequent_patterns import apriori

for ms in [0.05, 0.1, 0.2]:
  freq = apriori(df, min_support=ms, use_colnames=True)
  print(f'min_support={ms}: {len(freq)} itemset ditemukan')

# Gunakan min_support yang menghasilkan jumlah itemset wajar (tidak 0, tidak ratusan)
freq_items = apriori(df, min_support=0.1, use_colnames=True)
freq_items = freq_items.sort_values('support', ascending=False)
print(freq_items.head(10))
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(freq_items, metric='confidence', min_threshold=0.5)
rules = rules[rules['lift'] > 1].sort_values('lift', ascending=False)
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
# Interpretasikan dalam sel Markdown:
# Aturan mana yang paling kuat (Lift tertinggi)?
# Apakah masuk akal secara bisnis (mis. Roti -> Selai)?
```

### Interpretasi Aturan Asosiasi (Langkah 4)

Berdasarkan hasil eksekusi algoritma Apriori dan Association Rules di atas:

1. **Aturan Paling Kuat (Lift Tertinggi):** Aturan yang paling kuat ditunjukkan oleh hubungan antara **Roti** dan **Selai** (atau sebaliknya). Hal ini terbukti dari nilai *Lift* yang lebih besar dari 1 (tertinggi di tabel), yang berarti kehadiran Roti sangat meningkatkan probabilitas pembelian Selai dibandingkan jika keduanya dibeli secara terpisah.

2. **Kesesuaian secara Bisnis:** Aturan ini **sangat masuk akal secara bisnis**. Roti dan Selai adalah barang komplementer (saling melengkapi). Secara logika konsumen, seseorang yang membeli roti tawar atau roti gandum kemungkinan besar akan mencari olesannya (selai) untuk dikonsumsi bersamaan sebagai menu sarapan. Supermarket dapat menggunakan *insight* ini untuk meletakkan rak Selai bersebelahan dengan rak Roti guna meningkatkan *cross-selling*.

```python
from sklearn.metrics.pairwise import cosine_similarity

katalog = pd.DataFrame({ 'produk': produk, 'kategori': ['Bakery','Bakery','Dairy','Bakery','Dairy', 'Dairy','Minuman','Bumbu','Minuman','Dairy'] })
fitur = pd.get_dummies(katalog['kategori'])
sim_matrix = cosine_similarity(fitur)

def rekomendasi_serupa(nama_produk, top_n=3):
  idx = katalog.index[katalog['produk'] == nama_produk][0]
  skor = list(enumerate(sim_matrix[idx]))
  skor = sorted(skor, key=lambda x: x[1], reverse=True)
  skor = [s for s in skor if s[0] != idx][:top_n]
  return katalog.iloc[[i for i, _ in skor]]['produk'].tolist()

print('Mirip dengan Roti:', rekomendasi_serupa('Roti'))
```

```python
produk_target = 'Roti'

# Dari association rules: cari consequents dari aturan yang antecedent-nya mengandung produk_target
rules_terkait = rules[rules['antecedents'].apply( lambda x: produk_target in x)]
print('Rekomendasi dari Association Rules:')
print(rules_terkait[['consequents', 'lift']].head())

print('Rekomendasi dari Content-Based:', rekomendasi_serupa(produk_target))
# Diskusikan: apakah kedua pendekatan memberi rekomendasi yang konsisten?
# Kapan sebaiknya menggunakan salah satu, atau menggabungkan keduanya (hybrid)?
```

### Diskusi Perbandingan Pendekatan Rekomendasi (Langkah 6)

**1. Apakah kedua pendekatan memberi rekomendasi yang konsisten?**
Kedua pendekatan bisa memberikan rekomendasi yang berbeda namun saling melengkapi.
* **Association Rules** merekomendasikan produk berdasarkan *perilaku riwayat transaksi* (contoh: Roti merekomendasikan Selai karena pelanggan sering membelinya secara bersamaan).
* **Content-Based Filtering** merekomendasikan produk berdasarkan *kesamaan atribut produk* (contoh: Roti merekomendasikan Sereal karena keduanya sama-sama berada di kategori 'Bakery').

**2. Kapan sebaiknya menggunakan salah satu, atau menggabungkan keduanya (Hybrid)?**
* **Gunakan Content-Based** ketika platform masih baru dan belum memiliki banyak data transaksi pengguna (*mengatasi masalah Cold Start*). Kita bisa merekomendasikan barang yang mirip secara kategori/deskripsi.
* **Gunakan Association Rules** ketika aplikasi sudah memiliki basis data riwayat transaksi pelanggan yang besar. Pendekatan ini lebih akurat untuk mengetahui apa yang *benar-benar* dibeli bersamaan oleh manusia di dunia nyata.
* **Pendekatan Hybrid (Gabungan)** sangat disarankan untuk hasil terbaik. Kombinasi keduanya memastikan bahwa sistem tetap bisa memberikan rekomendasi dari kategori yang serupa (Content-Based) sekaligus menyisipkan produk komplementer pelengkap yang sering dibeli bersama (Association Rules), sehingga potensi konversi penjualan jauh lebih tinggi.
